```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller, acf 
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf


warnings.filterwarnings("ignore", category=FutureWarning)
plt.rcParams["figure.figsize"] = (11, 3.4)
```

## Part A. Step 1. Load the data and read the header

```python
URL = "https://gml.noaa.gov/webdata/ccgg/trends/co2/co2_mm_mlo.csv"
raw = pd.read_csv(URL, comment="#")
raw.tail()
```

```python
raw.index = pd.to_datetime(raw[["year", "month"]].assign(day=1))
raw = raw.asfreq("MS") # one row per month, dated at the start of the month
co2 = raw["average"] # monthly mean C02, in ppm
co2.plot(title="Monthly mean CO$_2$ at Mauna Loa (ppm)")
plt.show()
```

```python
train = co2.loc["1975-01":"2025-12"]
test = co2. loc["2026-01":"2026-09"]
print(len(train), "training months,", len(test), "test months")
raw. loc["1975":"2025"].query("sdev < 0") [["average", "ndays", "sdev"]]
```

Two interpolated months out of 612 (December 1975 and April 1984). We keep them and say so.

## Step 2 - Plot it and decide: log or no log?

```python
train.plot(title="Training window, 1975-2025")
plt.show()

seasonal = (raw["average"] - raw["deseasonalized"]).loc["1975":"2025"] # NOAA's seasonal component, ppm
swing = seasonal.groupby(seasonal.index.year).max() - seasonal.groupby(seasonal.index.year).min()
level = train.groupby(train.index.year).mean()
plt.scatter(level, swing, label="observed swing")
plt.plot(level, swing[1975] * level / level[1975], color="C1", label="if it grew in proportion to the level")
plt.ylim(0, None)
plt.xlabel("yearly mean (ppm)")
plt.ylabel("seasonal swing (ppm)")
plt.legend()
plt.show()
print(f"level: {level[1975]:.0f} -> {level[2025]:.0f} ppm ({level[2025] / level[1975] - 1:+.0%})")
print(f"swing: {swing[1975]:.2f} -> {swing[2025]:.2f} ppm ({swing[2025] / swing[1975] - 1:+.0%})")
```

## Step 3 - Choose the differences, then test the result

```python
print ("ADF, raw series (constant + trend): p =", round(adfuller(train, regression="ct")[1], 3))
plot_acf(train.diff().dropna(), lags=48, title="ACF after one regular difference")
plt.show()
print("ACF at lags 12, 24, 36, 48:", acf(train.diff().dropna(), nlags=48)[[12, 24, 36, 48]].round(2))
```

The raw series: p = 0.995, so we cannot reject a unit root, hence d = 1. After that one difference the ACF still has huge bars at 12, 24, 36 and 48 that fade very slowly (0.92 down to 0.86): the seasonal pattern survives, hence D = 1. Same choices as with the passengers.

```python
z = train.diff().diff(12).dropna() # (1 - B)(1 - B^12) X_t
print("ADF, z: p =", f"{adfuller(z)[1]:.1e}")
z.plot(title="z: this month's change minus the same month's change one year earlier")
plt.show()
```

## Step 4 - read p, q, P, Q from the ACF and the PACF

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 3.5))
plot_acf(z, lags=36, ax=ax[0])
plot_pacf(z, lags=36, ax=ax[1], method="ywm")
plt.show()


acf_z = acf(z, nlags=36)
lags = [1, 2, 3, 11, 12, 13, 14, 24, 36]
pd.Series(acf_z[lags], index=lags).round(2)
```

For this model the theoretical ACF is zero everywhere except at lags 1, 11, 12 and 13, with p11=p13 = θΘ/ ((1+ θ^2)(1 + Θ^2).

```python
fit = sm.tsa.SARIMAX(train, order=(0, 1, 1), seasonal_order=(0, 1, 1, 12)).fit(disp=False)
print(fit.summary().tables[1])

theta, Theta = fit.params["ma.L1"], fit.params["ma.S.L12"]
print("ACF the model predicts at lags 11 and 13:", round(theta * Theta / ((1 + theta**2) * (1 + Theta**2)), 2))
print("ACF in the data at lags 11 and 13:", round(acf_z[11], 2), round(acf_z[13], 2))
```

```python
candidates = {
"airline (0,1,1)(0,1,1)": ((0, 1, 1), (0, 1, 1, 12)),
"+ AR(1)": ((1, 1, 1), (0, 1, 1, 12)),
"+ MA(2), a = 2": ((0, 1, 2), (0, 1, 1, 12)),
"+ seasonal AR(1)": ((0, 1, 1), (1, 1, 1, 12)),
"AR version (1,1,0)(1,1,0)": ((1, 1, 0), (1, 1, 0, 12)),
}
rows = []
for name, (order, seasonal_order) in candidates.items():
    f= sm.tsa.SARIMAX(train, order=order, seasonal_order=seasonal_order).fit(disp=False) 
    k= len(f.params) -1 # ARMA terms (sigma2 is not one)
    lb_p = acorr_ljungbox(f.resid.iloc[13:], lags=[24], model_df=k)["lb_pvalue"].iloc[0]
    rows.append({"model": name, "AIC": round(f.aic, 1), "BIC": round(f.bic, 1), "Ljung-Box p": round(lb_p, 3)})
pd.DataFrame(rows).sort_values("BIC")
```
