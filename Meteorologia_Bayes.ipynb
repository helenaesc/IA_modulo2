# Bayes

```python
# Librerias y datos
import pandas as pd

# Cargar dataset
df = pd.read_csv(
    "meteorologia/meteorología_2025.csv",
    skiprows=9,
    low_memory=False
)

df.head()
```

```python
# Selección de estación y parametros

# Estación
uax = df[df["id_station"] == "UAX"]

# Parámetros
uax = uax[uax["id_parameter"].isin(["RH", "WSP"])]

# Convertimos de formato largo a formato ancho
uax = uax.pivot(
    index="date",
    columns="id_parameter",
    values="valor"
).reset_index()

# Quitamos las filas donde falten los parámetros
uax = uax.dropna(subset=["RH", "WSP"])

uax.head()
```

```python
# Definimos eventos 
humedad_alta = uax["RH"] >= 70
viento_tranquilo = uax["WSP"] <= 1

# Intersección de ambos eventos
ambos = humedad_alta & viento_tranquilo
```

## Probabilidades

```python
# Probabilidades simples
p_humedad_alta = humedad_alta.mean()
p_viento_tranquilo = viento_tranquilo.mean()
```

```python
# Probabilidades condicionales
p_hum_dado_viento = ambos.sum() / viento_tranquilo.sum()
p_viento_dado_hum = ambos.sum() / humedad_alta.sum()
```

```python
# Resultados
print(
    "P(Humedad alta):",
    round(p_humedad_alta, 4),
    "=",
    round(p_humedad_alta * 100, 2),
    "%"
)

print(
    "P(Viento tranquilo):",
    round(p_viento_tranquilo, 4),
    "=",
    round(p_viento_tranquilo * 100, 2),
    "%"
)

print(
    "P(Hum alta | Viento tranquilo):",
    round(p_hum_dado_viento, 4),
    "=",
    round(p_hum_dado_viento * 100, 2),
    "%"
)

print(
    "P(Viento tranquilo | Hum alta):",
    round(p_viento_dado_hum, 4),
    "=",
    round(p_viento_dado_hum * 100, 2),
    "%"
)
```
